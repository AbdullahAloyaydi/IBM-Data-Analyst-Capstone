### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Example aggregated data
data = {
    'AgeGroup': ['18–25', '26–35', '36–45', '46+'],
    'JobSatPoints_6': [3.2, 3.8, 4.1, 4.0],
    'JobSatPoints_7': [2.1, 2.6, 3.0, 3.2]
}

df = pd.DataFrame(data)

# Plot stacked bar chart
plt.figure()
plt.bar(df['AgeGroup'], df['JobSatPoints_6'], label='JobSatPoints_6')
plt.bar(
    df['AgeGroup'],
    df['JobSatPoints_7'],
    bottom=df['JobSatPoints_6'],
    label='JobSatPoints_7'
)

plt.xlabel('Age Groups')
plt.ylabel('Median Job Satisfaction Score')
plt.title('Stacked Chart of Job Satisfaction Across Age Groups')
plt.legend()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Example aggregated data
data = {
    'EmploymentStatus': ['Full-time', 'Part-time', 'Contract', 'Unemployed'],
    'JobSatPoints_6': [4.1, 3.5, 3.8, 2.9],
    'JobSatPoints_7': [3.2, 2.4, 2.7, 1.8]
}

df = pd.DataFrame(data)

# Create stacked bar chart
plt.figure()
plt.bar(
    df['EmploymentStatus'],
    df['JobSatPoints_6'],
    label='JobSatPoints_6'
)

plt.bar(
    df['EmploymentStatus'],
    df['JobSatPoints_7'],
    bottom=df['JobSatPoints_6'],
    label='JobSatPoints_7'
)

plt.xlabel('Employment Status')
plt.ylabel('Median Job Satisfaction Score')
plt.title('Stacked Chart of Job Satisfaction by Employment Status')
plt.legend()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
##Write your code here
import matplotlib.pyplot as plt

# Group data by Employment Status and calculate median values
grouped_df = df.groupby('EmploymentStatus')[[
    'JobSatPoints_6',
    'JobSatPoints_7'
]].median()

# Create stacked bar chart
plt.figure()

plt.bar(
    grouped_df.index,
    grouped_df['JobSatPoints_6'],
    label='JobSatPoints_6'
)

plt.bar(
    grouped_df.index,
    grouped_df['JobSatPoints_7'],
    bottom=grouped_df['JobSatPoints_6'],
    label='JobSatPoints_7'
)

plt.xlabel('Employment Status')
plt.ylabel('Median Job Satisfaction Score')
plt.title('Stacked Chart of Job Satisfaction by Employment Status')
plt.legend()
plt.xticks(rotation=30)
plt.show()




```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Example structure (conceptual)
data = {
    'AgeGroup': ['18-24', '18-24', '25-34', '25-34', '35-44', '35-44'],
    'Database': ['MySQL', 'PostgreSQL', 'MySQL', 'MongoDB', 'PostgreSQL', 'MongoDB'],
    'Count': [40, 30, 50, 35, 45, 25]
}

df_example = pd.DataFrame(data)

# Pivot to prepare for stacked chart
pivot_df = df_example.pivot(
    index='AgeGroup',
    columns='Database',
    values='Count'
)

# Convert counts to proportions
pivot_df = pivot_df.div(pivot_df.sum(axis=1), axis=0)

# Plot stacked chart
pivot_df.plot(kind='bar', stacked=True)
plt.ylabel('Proportion')
plt.title('Stacked Chart of Preferred Databases by Age Group')
plt.legend(title='Database')
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
##Write your code here
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create a cross-tabulation of Job Satisfaction and Employment Status
crosstab = pd.crosstab(
    df['JobSatPoints_6'],
    df['EmploymentStatus']
)

# Step 2: Convert counts to proportions
crosstab_prop = crosstab.div(crosstab.sum(axis=1), axis=0)

# Step 3: Plot stacked bar chart
crosstab_prop.plot(
    kind='bar',
    stacked=True
)

plt.xlabel('Job Satisfaction Level (JobSatPoints_6)')
plt.ylabel('Proportion of Employment Types')
plt.title('Stacked Chart of Employment Type by Job Satisfaction Level')
plt.legend(title='Employment Status')
plt.show()
```

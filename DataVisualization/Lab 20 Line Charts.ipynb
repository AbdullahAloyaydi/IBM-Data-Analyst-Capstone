### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt


grouped_df = (
    df.groupby('Age')['ConvertedCompYearly']
    .median()
    .dropna()
)

#  Line Chart
plt.figure()
plt.plot(
    grouped_df.index,
    grouped_df.values,
    marker='o'
)

plt.xlabel('Age Group')
plt.ylabel('Median ConvertedCompYearly')
plt.title('Line Chart of Median ConvertedCompYearly by Age Group')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()


```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
## Write your code here
import matplotlib.pyplot as plt

# Filter ages between 25 and 45 using text matching
filtered_df = df[
    df['Age'].str.contains('25|34|35|44', na=False)
]

grouped_df = (
    filtered_df.groupby('Age')['ConvertedCompYearly']
    .median()
    .dropna()
)

# Plot line chart
plt.figure()
plt.plot(
    grouped_df.index,
    grouped_df.values,
    marker='o'
)

plt.xlabel('Age Group')
plt.ylabel('Median ConvertedCompYearly')
plt.title('Line Chart of Median ConvertedCompYearly for Ages 25 to 45')
plt.grid(True)
plt.show()


```

#### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt


df_exp = df.copy()

df_exp['YearsCodePro'] = df_exp['YearsCodePro'].replace({
    'Less than 1 year': 0,
    'More than 50 years': 50
})

df_exp['YearsCodePro'] = pd.to_numeric(df_exp['YearsCodePro'], errors='coerce')


grouped_df = (
    df_exp.groupby('YearsCodePro')['JobSatPoints_6']
    .median()
    .dropna()
    .sort_index()
)

#  Line Chart
plt.figure()
plt.plot(
    grouped_df.index,
    grouped_df.values,
    marker='o'
)

plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction (JobSatPoints_6)')
plt.title('Line Chart of Job Satisfaction by Experience Level')
plt.grid(True)
plt.show()
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
df = df.dropna(subset=['YearsCodePro', 'JobSat'])
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

bins = [0, 5, 10, 20, df['YearsCodePro'].max()]
labels = ['0-5 years', '5-10 years', '10-20 years', '20+ years']

df['ExperienceRange'] = pd.cut(
    df['YearsCodePro'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
median_jobsat = df.groupby('ExperienceRange')['JobSat'].median()
median_jobsat

import matplotlib.pyplot as plt

median_jobsat.plot(kind='bar')
plt.title('Median Job Satisfaction by Experience Range')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
## Write your code here
df = df.dropna(subset=['RemoteWork', 'Employment'])
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.countplot(x='RemoteWork', data=df)
plt.title('Distribution of Remote Work Preferences')
plt.xlabel('Remote Work Preference')
plt.ylabel('Number of Respondents')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.countplot(x='Employment', hue='RemoteWork', data=df)
plt.title('Remote Work Preferences by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.show()

remote_employment_ct = pd.crosstab(
    df['Employment'],
    df['RemoteWork']
)

remote_employment_ct

remote_employment_pct = remote_employment_ct.div(
    remote_employment_ct.sum(axis=1),
    axis=0
) * 100

remote_employment_pct
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1. Clean required columns
# -----------------------------
df = df.dropna(subset=['Country', 'LanguageHaveWorkedWith'])

# -----------------------------
# 2. Inspect available countries (important)
# -----------------------------
print("Top countries in the dataset:")
print(df['Country'].value_counts().head(10))

# -----------------------------
# 3. Select region / country
# (Change country name if needed)
# -----------------------------
selected_country = 'United States'

region_df = df[
    df['Country']
    .astype(str)
    .str.contains(selected_country, case=False, na=False)
]

print(f"\nRecords found for {selected_country}: {len(region_df)}")

# Safety check
if region_df.empty:
    raise ValueError("No data available for the selected region.")

# -----------------------------
# 4. Split and explode languages
# -----------------------------
region_df = region_df.copy()

region_df['LanguageHaveWorkedWith'] = (
    region_df['LanguageHaveWorkedWith']
    .astype(str)
    .str.split(';')
)

region_df = region_df.explode('LanguageHaveWorkedWith')

# -----------------------------
# 5. Count top languages
# -----------------------------
top_languages = (
    region_df['LanguageHaveWorkedWith']
    .value_counts()
    .head(10)
)

print("\nTop Programming Languages:")
print(top_languages)

# -----------------------------
# 6. Bar Plot visualization
# -----------------------------
plt.figure(figsize=(12, 7))
top_languages.plot(kind='bar')
plt.title(f'Top Programming Languages in {selected_country}')
plt.xlabel('Programming Language')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# -----------------------------
# 7-(Optional) Heatmap for multiple countries
# -----------------------------
top_n_languages = 5
top_languages_global = (
    df['LanguageHaveWorkedWith']
    .astype(str)
    .str.split(';')
    .explode()
    .value_counts()
    .head(top_n_languages)
    .index
)

heatmap_df = df.copy()
heatmap_df['LanguageHaveWorkedWith'] = (
    heatmap_df['LanguageHaveWorkedWith']
    .astype(str)
    .str.split(';')
)

heatmap_df = heatmap_df.explode('LanguageHaveWorkedWith')

heatmap_data = pd.crosstab(
    heatmap_df['Country'],
    heatmap_df['LanguageHaveWorkedWith']
)[top_languages_global]

# Reduce to top 10 countries for clarity
heatmap_data = heatmap_data.loc[
    heatmap_data.sum(axis=1).sort_values(ascending=False).head(10).index
]

plt.figure(figsize=(14, 8))
sns.heatmap(heatmap_data, cmap='Blues')
plt.title('Top Programming Languages by Region')
plt.xlabel('Programming Language')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt

df_edu = df.dropna(subset=['EdLevel', 'Employment']).copy()

def simplify_employment(emp):
    if 'full-time' in emp:
        return 'Full-time'
    elif 'part-time' in emp:
        return 'Part-time'
    elif 'Independent contractor' in emp or 'self-employed' in emp:
        return 'Self-employed'
    elif 'Student' in emp:
        return 'Student'
    elif 'Retired' in emp:
        return 'Retired'
    elif 'Not employed' in emp:
        return 'Unemployed'
    else:
        return 'Other'

df_edu['EmploymentSimple'] = df_edu['Employment'].astype(str).apply(simplify_employment)
edu_emp_ct = pd.crosstab(
    df_edu['EdLevel'],
    df_edu['EmploymentSimple']
)

edu_emp_ct
edu_emp_pct = edu_emp_ct.div(
    edu_emp_ct.sum(axis=1),
    axis=0
) * 100

edu_emp_pct
edu_emp_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7)
)

plt.title('Employment Type by Educational Level (Percentage)')
plt.xlabel('Educational Level')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

```

### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
## Write your code here
df.shape
df.columns
df.info()
df.describe()
df.describe(include='all')
```

#### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
df.isnull().sum()

categorical_cols = ['Employment', 'RemoteWork', 'EdLevel']

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

numerical_cols = ['YearsCodePro', 'JobSat', 'JobSatPoints_1']

for col in numerical_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
df = df.dropna(subset=['Employment', 'EdLevel', 'YearsCodePro'])
df.isnull().sum()

```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
## Write your code here
df['Employment'].value_counts()
df['JobSat'].value_counts().sort_index()
df['YearsCodePro'].value_counts().sort_index()
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns

jobsat_df = df.dropna(subset=['JobSat'])

jobsat_counts = jobsat_df['JobSat'].value_counts().sort_index()

plt.figure(figsize=(7, 7))
plt.pie(
    jobsat_counts,
    labels=jobsat_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Distribution of Job Satisfaction (JobSat)')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.kdeplot(
    data=jobsat_df,
    x='JobSat',
    fill=True
)
plt.title('KDE Plot of Job Satisfaction')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt

df_lang = df.dropna(
    subset=['LanguageHaveWorkedWith', 'LanguageWantToWorkWith']
).copy()
worked_lang = (
    df_lang['LanguageHaveWorkedWith']
    .astype(str)
    .str.split(';')
    .explode()
    .value_counts()
)
wanted_lang = (
    df_lang['LanguageWantToWorkWith']
    .astype(str)
    .str.split(';')
    .explode()
    .value_counts()
)
top_languages = worked_lang.head(10).index

comparison_df = pd.DataFrame({
    'Worked With': worked_lang[top_languages],
    'Want To Work With': wanted_lang[top_languages]
}).fillna(0)

comparison_df
comparison_df.plot(
    kind='bar',
    figsize=(14, 7)
)

plt.title('Programming Languages: Worked With vs Want To Work With')
plt.xlabel('Programming Language')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Category')
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df_remote = df.dropna(subset=['Country', 'RemoteWork']).copy()
top_countries = (
    df_remote['Country']
    .value_counts()
    .head(10)
    .index
)

df_remote = df_remote[df_remote['Country'].isin(top_countries)]
plt.figure(figsize=(14, 7))
sns.countplot(
    data=df_remote,
    x='Country',
    hue='RemoteWork'
)

plt.title('Remote Work Distribution by Region')
plt.xlabel('Country / Region')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Remote Work Type')
plt.tight_layout()
plt.show()
remote_ct = pd.crosstab(
    df_remote['Country'],
    df_remote['RemoteWork']
)
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
## Write your code here
import pandas as pd

# -----------------------------
# 1. Select required columns
# -----------------------------
df_corr = df[['JobSat', 'YearsCodePro']].copy()

# -----------------------------
# 2. Remove missing values
# -----------------------------
df_corr = df_corr.dropna()

# -----------------------------
# 3. Convert columns to numeric
# -----------------------------
df_corr['JobSat'] = pd.to_numeric(df_corr['JobSat'], errors='coerce')
df_corr['YearsCodePro'] = pd.to_numeric(df_corr['YearsCodePro'], errors='coerce')

# Remove NaN created after conversion
df_corr = df_corr.dropna()

# -----------------------------
# 4. Pearson Correlation
# -----------------------------
pearson_corr = df_corr['JobSat'].corr(
    df_corr['YearsCodePro'],
    method='pearson'
)

print("Pearson Correlation:", pearson_corr)

# -----------------------------
# 5. Spearman Correlation (without SciPy)
# -----------------------------
job_sat_rank = df_corr['JobSat'].rank()
experience_rank = df_corr['YearsCodePro'].rank()

spearman_corr = job_sat_rank.corr(experience_rank)

print("Spearman Correlation:", spearman_corr)
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
## Write your code here
import pandas as pd
import matplotlib.pyplot as plt


df_ct = df.dropna(subset=['Employment', 'EdLevel']).copy()
employment_education_ct = pd.crosstab(
    df_ct['EdLevel'],
    df_ct['Employment']
)

employment_education_ct
employment_education_ct.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7)
)

plt.title('Employment Status by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Employment Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
